```python
import pandas as pd
import matplotlib.pyplot as plt

cols = ["unit", "cycle", "setting_1", "setting_2", "setting_3"] + [f"s_{i}" for i in range(1, 22)]
train = pd.read_csv("../data/train_FD001.txt", sep=r"\s+", header=None, names=cols)
train.head()
```

```python
moteur1 = train[train["unit"] == 1]
moteur1[["cycle", "s_2", "s_3", "s_4"]]
```

```python
plt.plot(moteur1["cycle"], moteur1["s_4"])
plt.xlabel("cycle (numéro du vol)")
plt.ylabel("capteur s_4")
plt.title("Moteur 1")
plt.show()
```

```python
capteurs = [f"s_{i}" for i in range(1, 22)]
moteur1.plot(x="cycle", y=capteurs, subplots=True, layout=(7, 3), figsize=(15, 18))
plt.show()
```
